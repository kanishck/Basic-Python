```python
# Hollow Triangle

n = 4
for row in range(1,n+1):
    for col in range(1, 2*n):
        if row == n or row + col == n+1 or col - row == n-1:
            print('*', end = ' ')
        else:
            print(' ', end =' ')
    print()
```

```python
# Hollow Square
n=7
for row in range(n):
    for col in range(n):
        if col==n-1 or row==n-1 or row==0 or col==0:
            print("*", end = ' ')
        else:
            print(' ', end = ' ')
    print()
```
