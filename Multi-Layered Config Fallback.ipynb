```python
# Initial Data

default_config = {
"theme": "light",
"timeout": 30,
"precision": "high"
}

environment_variables = {
"theme": "dark"
}
user_override = {
"timeout": 60
}
requested_keys = [
"timeout",
"theme",
"precision",
"api_key"
]

# Your Task

# 1. Create an empty dictionary named resolved_settings .
# 2. Iterate through each key in requested_keys .
# 3. Search for the value using the following priority:
#               user_override
#               environment_variables
#               default_config
# 4. If the key does not exist anywhere, assign -1 .
# 5. Store the resolved value in resolved_settings .
# 6. Print all available keys from default_config using a dictionary keys method.
# 7. Check whether "high" exists inside the values of default_config .
```

```python
resolved_settings = {}

for key in requested_keys:
    if key in user_override:
        resolved_settings[key] = user_override[key]
    elif key in environment_variables:
        resolved_settings[key] = environment_variables[key]
    elif key in default_config:
        resolved_settings[key] = default_config[key]
    else:
        resolved_settings[key] = -1
```

```python
for keys in default_config.keys():
    print(keys)
```

```python
for key,value in default_config.items():
    if value == 'high':
        print('The high exists')
        break
```
