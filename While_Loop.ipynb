```python
# Print numbers from 1 to N , but replace multiples of 3 with "Fizz" .

N = int(input("Enter N : "))

i=1

while i <= N :
    if i%3 == 0:
        print("Fizz")
        i += 1
        continue
    print(i)
    i += 1
```

```python
# Print numbers from 1 to N , but:
# Multiples of 3 → "Fizz"
# Multiples of 5 → "Buzz"
# Multiples of both → "FizzBuzz"

N= int(input("Enter N : "))

i=1

while i<=N :
    if i%3==0 and i%5==0:
        print("FizzBuzz")
        i+=1
        continue
    elif i%3 == 0:
        print("Fizz")
        i+=1
        continue
    elif i%5 == 0:
        print("Buzz")
        i+=1
        continue
    else:
        print(i)
    i+=1
```

```python
# Find the sum of all numbers between 1 and N that are divisible by 7 .

N = int(input("Enter N : "))
sum = 0
i=7

while i<=N :
    sum += i
    i+=7
print(sum)
```

```python
# Print all numbers between 1 and N whose square is less than 1000 .

N = int(input("Enter N : "))

i=1

while i*i < 1000 and i<=N :
    print(i)
    i+=1
```

```python
# Find the first number greater than N that is divisible by both 4 and 6

N = int(input("Enter N : "))

i=N

while True:
    if i%4 == 0 and i%6 == 0:
        print(i)
        break
    i+=1
```

```python
# Find the smallest multiple of 13 greater than a given number.

N = int(input("Enter N : "))
i=N
while True:
    if i%13 == 0 :
        print(i)
        break
    i += 1
```
