```python
# Print numbers from 1 to N , but replace multiples of 3 with "Fizz" .

N = int(input("Enter N : "))

i=1

while i <= N :
    if i%3 == 0:
        print("Fizz")
        i += 1
        continue
    print(i)
    i += 1
```

```python
# Print numbers from 1 to N , but:
# Multiples of 3 → "Fizz"
# Multiples of 5 → "Buzz"
# Multiples of both → "FizzBuzz"

N= int(input("Enter N : "))

i=1

while i<=N :
    if i%3==0 and i%5==0:
        print("FizzBuzz")
        i+=1
        continue
    elif i%3 == 0:
        print("Fizz")
        i+=1
        continue
    elif i%5 == 0:
        print("Buzz")
        i+=1
        continue
    else:
        print(i)
    i+=1
```

```python
# Check whether a number contains duplicate consecutive digits.

N = int(input("Enter N : "))

temp = N%10
N//=10

while N!=0 :
    digit = N%10
    if digit == temp:
        print("duplicate number found -> ",digit)
        break
    else:
        temp = digit
        N//=10
if N==0 :
    print("No duplicates found")
```
