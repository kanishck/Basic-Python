```python
# print the key which has the highest value
dic = {'a':3,'b':1,'c':5}
largest = 0
target=''
for key in dic:
    if dic[key]>largest:
        largest = dic[key]
        target = key

print(target)
```

```python
d = {'a':3, 'b':1, 'c':5}
new = {}
for key,value in d.items():
    new[value] = key
print(new)
```

```python
d = 'aabbcccdddd'
dic = {}

for i in range(len(d)):
    if d[i] in dic:
        dic[d[i]] += 1
    else:
        dic[d[i]] = 1

largest = 0
largest_key = ''
for key, value in dic.items():
    if dic[key] > largest:
        largest = dic[key]
        largest_key = key
print(largest_key)
```
