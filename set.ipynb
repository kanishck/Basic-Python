```python
sent = "python is easy python is powerful"
li = sent.split()
seen = set()
for word in li:
    if word not in seen:
        print(word,end=' ')
        seen.add(word)
```

```python
s = 'apple banana orange mango umbrella grape'.split()
vowels = 'aeiou'
li = set()
for word in s:
    if word[0] not in vowels:
        li.add(word)
print(li)
```

```python
li = [1,2,2,3,4]
l1 = [4,3,2,1]
is_present = True
for el in li:
    if el not in l1:
        is_present = False

if is_present:
    print('Yes')
else:
    print('No')
```
