```python
# Initial Data

inventory = {
"Laptop": 10,
"Mouse": 0,
"Keyboard": 5,
"Monitor": 0,
"HDMI Cable": 15
}

# Note: "HDMI Cable" is the last inserted item.

# Your Task

# 1. Create an empty list named removed_items .
# 2. Since dictionary size cannot be modified while iterating, first create a separate list of dictionary
# keys and iterate over that list.
# 3. If an item's quantity is 0 , remove it from the dictionary using an appropriate dictionary method.
# 4. Store the removed item name in removed_items .
# 5. After removing all zero-quantity items, remove the last inserted key-value pair using a suitable
# dictionary method.
# 6. Store that removed key-value pair in a variable named liquidated_item .
# 7. Print:
# Updated inventory
# removed_items
# liquidated_item
```

```python
removed_items = []
keys=[]
for key in inventory:
    keys.append(key)
keys
```

```python
for key in keys:
    if inventory[key] == 0:
        removed_items.append(key)
        inventory.pop(key)

inventory
```

```python
liquidated_item = inventory.popitem()
```

```python
print(inventory)
print(removed_items)
print(liquidated_item)
```
