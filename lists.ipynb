```python
sum_even=0
sum_odd=0

for el in li:
    if el%2==0:
        sum_even+=el
    else:
        sum_odd+=el
print(sum_odd)
print(sum_even)
```

```python
li = input("Enter the list items separated by spaces: ").split()

largest = int(li[0])

for el in li:
    if int(el)>largest:
        largest = int(el)

print(largest)
```
