```python
# Read how many numbers will be entered
n = int(input("Enter N: "))

found = False

for i in range(n):

    num = int(input("Enter number: "))

    # Stop as soon as a negative number is found
    if num < 0:
        print("First Negative Number:", num)
        found = True
        break

# If loop finished without finding a negative number
if found == False:
    print("No Negative Number Found")
```

```python
# Number of subjects
n = int(input("Enter number of subjects: "))

failed = False

for i in range(n):

    marks = int(input("Enter marks: "))

    # Fail immediately if marks are below 33
    if marks < 33:
        print("Result: Fail")
        failed = True
        break

# Executed only if no subject had failing marks
if failed == False:
    print("Result: Pass")
```

```python
# Number of students
n = int(input("Enter number of students: "))

total = 0
count = 0

for i in range(n):

    marks = int(input("Enter marks: "))

    # Ignore invalid marks
    if marks < 0:
        continue

    total += marks
    count += 1

# Avoid division by zero
if count > 0:
    print("Average =", total / count)
else:
    print("No valid marks entered")
```

```python
# Number of students
n = int(input("Enter number of students: "))

absent_count = 0

for i in range(n):

    attendance = input("Enter attendance (P/A): ")

    # Ignore invalid input
    if attendance != "P" and attendance != "A":
        continue

    # Count absentees
    if attendance == "A":
        absent_count += 1

    # Stop when 10 absentees are found
    if absent_count == 10:
        print("10 absentees found.")
        break
```

```python
n = int(input("Enter number of vehicles: "))

total = 0

for i in range(n):

    toll = int(input("Enter toll amount: "))

    # Ignore vehicles that paid nothing
    if toll == 0:
        continue

    total += toll

    # Stop after exceeding the target
    if total > 50000:
        print("Collection exceeded ₹50,000")
        break
```

```python
runs = 0
wickets = 0

while True:

    score = int(input("Enter runs (-1 for wide, 0 for wicket): "))

    # Ignore wide balls
    if score == -1:
        continue

    # Wicket
    if score == 0:
        wickets += 1

        if wickets == 10:
            print("All out!")
            break

    else:
        runs += score

print("Total Runs:", runs)
```

```python
budget = int(input("Enter budget: "))
n = int(input("Enter number of products: "))

total = 0

for i in range(n):

    price = int(input("Enter product price: "))

    # Ignore products with price 0
    if price == 0:
        continue

    total += price

    # Stop if budget is exceeded
    if total > budget:
        print("Budget exceeded.")
        break

print("Final Bill:", total)
```

```python
n = int(input("Enter number of days: "))

for i in range(n):

    profit = int(input("Enter profit/loss: "))

    # Ignore zero-profit days
    if profit == 0:
        continue

    # Stop if loss exceeds ₹20,000
    if profit < -20000:
        print("Huge loss encountered.")
        break

    print("Profit/Loss:", profit)
```

```python
n = int(input("Enter number of ratings: "))

one_star_count = 0

for i in range(n):

    rating = int(input("Enter rating (1-5): "))

    # Ignore invalid ratings
    if rating < 1 or rating > 5:
        continue

    # Count one-star ratings
    if rating == 1:
        one_star_count += 1

    # Stop after three one-star ratings
    if one_star_count == 3:
        print("Three 1-star ratings received.")
        break
```
