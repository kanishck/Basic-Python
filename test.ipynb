```python
# Q11

li = [4,5,1,2,1,2,4]
for i in range(len(li)):
    repeating = False
    for j in range(len(li)):
        if li[i] == li[j] and i!=j:
            repeating = True
            break
    if repeating == False:
        non_repeat = li[i]
        break

    print(non_repeat)
```

```python
#Q13

n = int(input("Enter n : "))

for row in range(n):
    for col in range(1, 2*n):
        if row+col == n or col == n+row :
            print("*", end='')
        else:
            print(" ", end='')
    print()


for row in range(2,1+ n):
    for col in range(1,2*n):
        if row==col or col == 2*n-row :
            print("*", end='')
        else:
            print(" ", end='')
    print()
```
