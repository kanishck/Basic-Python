```python
# Hollow Triangle

n = 4
for row in range(1,n+1):
    for col in range(1, 2*n):
        if row == n or row + col == n+1 or col - row == n-1:
            print('*', end = ' ')
        else:
            print(' ', end =' ')
    print()
```

```python
# Hollow Triangle second method

n = 4
for row in range(1,n+1):
    for col in range(1, 2*n):
        if (row == n and col%2==1) or row + col == n+1 or col - row == n-1:
            print('*', end = ' ')
        else:
            print(' ', end =' ')
    print()
```
