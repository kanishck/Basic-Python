```python
# Initial Data

# Your Task

# 1. Create a dictionary named tracker using allowed_events where every event type starts
# with an empty set() .
# allowed_events = ["LOGIN", "LOGOUT", "PURCHASE"]
# raw_logs = [
# ("10:00", "LOGIN", "user_1"),
# ("10:01", "PURCHASE", "user_2"),
# ("10:02", "LOGIN", "user_1"), # Duplicate user for LOGIN
# ("10:03", "ERROR_404", "user_3"), # New event type
# ("10:04", "ERROR_404", "user_1") # Duplicate user for the new event
# ]

# 2. Traverse each log in raw_logs .
# 3. Extract the event_type and user_id .
# 4. Use a dictionary method that automatically creates a new empty set if the event type is not
# already present.
# 5. Add the user_id to the corresponding set.
# 6. After processing all logs, add the following status information using a dictionary batch update
# method:

# {"system_status": "healthy"}

# 7. Print the final tracker dictionary.
```

```python
allowed_events = ['LOGIN', 'LOGOUT', 'PURCHASE']

raw_logs = [("10:00", "LOGIN", "user_1"),
("10:01", "PURCHASE", "user_2"),
("10:02", "LOGIN", "user_1"), # Duplicate user for LOGIN
("10:03", "ERROR_404", "user_3"), # Unlisted event type: handle dynamically
("10:04", "ERROR_404", "user_1") # Duplicate for the new unlisted event type
]

tracker= {event: set() for event in allowed_events}

tracker
```

```python
print(type(tracker['LOGIN']))
```

```python
for el in raw_logs:
    event = el[1]
    user = el[2]
    tracker.setdefault(event,set())
    eve_set = tracker[event]
    eve_set.add(user)
```

```python
tracker.update({"system_status": "healthy"})
```
